```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import minmax_scale
```

```python
x=df.drop(columns=['Restaurant ID','Restaurant Name','City','Address','Locality','Locality Verbose','Cuisines','Currency','Aggregate rating'])
y=df['Aggregate rating']
```

```python
x
```

Scaling

```python
sc=minmax_scale(x)
```

```python
X=pd.DataFrame(sc,columns=x.columns)
```

```python
x
```

```python
X
```

```python
#ns.pairplot(data=df)
```

train and test split

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,random_state=42,test_size=0.2)
```

linear regression

```python
lg=LinearRegression()
lg.fit(X_train,y_train)
```

```python
print(lg.score(X_train,y_train),lg.score(X_test,y_test))
```

```python
from sklearn.metrics import mean_absolute_error,mean_squared_error,root_mean_squared_error,r2_score,accuracy_score
```

```python
print("MSE:",mean_squared_error(y_test,lg.predict(X_test))*100)
print("MAE:",mean_absolute_error(y_test,lg.predict(X_test))*100)
print("RMSE:",np.sqrt(mean_squared_error(y_test,lg.predict(X_test)))*100)
print("r_2 score:",r2_score(y_test,lg.predict(X_test))*100)
```

```python
import tensorflow
import keras
```

```python
model=keras.Sequential([
    keras.layers.Dense(1000,input_shape=(20,),activation='relu'),
    keras.layers.Dense(700,activation='relu'),
    keras.layers.Dense(400,activation='relu'),
    keras.layers.Dense(100,activation='relu'),
    keras.layers.Dense(50,activation='relu'),
    keras.layers.Dense(20,activation='relu'),
    keras.layers.Dense(1)
])
model.compile(optimizer='adam',loss='mse',metrics=['mae'])
model.fit(X_train,y_train,epochs=100)
```

```python
rating = df.groupby('Cuisines').agg({'Aggregate rating':'mean','Restaurant ID':'count'})
rating.columns=['Average Rating','Count']
rating = rating[rating['Count']>=10]
rating.sort_values('Average Rating',ascending=False).head(10)
```

```python
from collections import defaultdict
rating_sum = defaultdict(float)
count = defaultdict(int)
for _, row in df.dropna(subset=['Cuisines']).iterrows():
    cuisines = [c.strip() for c in row['Cuisines'].split(',')]
    for cuisine in cuisines:
        rating_sum[cuisine] += row['Aggregate rating']
        count[cuisine] += 1
avg_rating = {
    c: rating_sum[c] / count[c]
    for c in count
}
result = (
    pd.Series(avg_rating)
      .sort_values(ascending=False)
)
print(result.head(10))
```
