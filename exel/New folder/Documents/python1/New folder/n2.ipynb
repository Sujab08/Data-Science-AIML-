```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv(r"C:\Users\kumar\Downloads\HR-Employee-Attrition.csv")
```

```python
df.info()
```

```python
plt.hist()
```

```python
df.groupby("EducationField")["Attrition"].value_counts().unstack().plot(kind="bar")
plt.ylabel("count")
```

```python
plt.xlabel("EducationField")
plt.title("Attrition rate vs EducationField")
plt.grid()
plt.show()
```

```python
bins = [18,25,35,45,55,60]
labels = ['18–25','26–35','36–45','46–55','56–60']
df['AgeGroup'] = pd.cut(df['Age'], bins=bins, labels=labels)
df.groupby('AgeGroup')['Attrition'].value_counts(normalize=True).unstack().plot(kind='bar', color=['lightgreen','salmon'])
plt.title('Attrition by Age Group')
```

```python
plt.xlabel('Age Group')
plt.ylabel('Proportion')
plt.show()
```

```python
df.groupby("JobRole")["Attrition"].value_counts().unstack().plot(kind="bar")
plt.ylabel("count")
plt.xlabel("department")
plt.title("Attrition rate vs JobRole")
plt.grid()
plt.show()
```

```python
df.groupby("Department")["Attrition"].value_counts().unstack().plot(kind="bar")
plt.ylabel("count")
plt.xlabel("department")
plt.title("Attrition rate vs Department")
plt.grid()
plt.show()
```

```python
df["Attrition"].value_counts().plot(kind="pie",autopct="%1.1f%%")
plt.legend()
plt.show()
```

```python
plt.scatter(df["MonthlyIncome"],df["JobLevel"],alpha=1,color="red")
plt.xlabel("monthly income")
plt.grid()
plt.ylabel("joblevel")
plt.show()
```

```python
plt.scatter(df["Age"],df["MonthlyIncome"],alpha=.3,color="red")
plt.xlabel("age")
plt.grid()
plt.ylabel("monthly salary")
plt.show()
```

```python
plt.hist(df["RelationshipSatisfaction"],color="red",edgecolor="black",bins=[1,2,3,4])
plt.xlabel("relationship distribution")
plt.ylabel("count")
plt.title("RELATIONSHIP DISTRIBUTION")
plt.grid()
plt.show()
```

```python
plt.hist(df["EnvironmentSatisfaction"],color="red",edgecolor="black",bins=[1,2,3,4])
plt.xlabel("enviromental distribution")
plt.ylabel("count")
plt.title("ENVIROMENTAL DISTRIBUTION")
plt.grid()
plt.show()
```

```python
df["TrainingTimesLastYear"].value_counts().sort_index().plot(kind="line",color="red",marker="o",linestyle="-.")
plt.grid()
plt.show()
```
