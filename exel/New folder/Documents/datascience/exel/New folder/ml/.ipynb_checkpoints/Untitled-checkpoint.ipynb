```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
d['genderr']=le.fit_transform(d['gender'])
d['race/ethnicityy']=le.fit_transform(d['race/ethnicity'])
#print(d[['genderr','race/ethnicityy']])
```

```python
d.head(10)
```

```python
from sklearn.preprocessing import StandardScaler
ss=StandardScaler()
ss.fit(d[['math score']])
d['math score_ss']=pd.DataFrame(ss.transform(d[['math score']]),columns=['x'])
```

```python
from sklearn.preprocessing import MinMaxScaler
mm=MinMaxScaler()
mm.fit(d[['math score']])
d['math score_mm']=pd.DataFrame(mm.transform(d[['math score']]),columns=['x'])
```

```python
d.head()
```

```python
from sklearn.model_selection import train_test_split
input=d[['math score','reading score']]
output=d['writing score']
x_train,x_test,y_train,y_test=train_test_split(input,output,test_size=0.25)
```

```python
plt.figure(figsize=[20,7])
plt.subplot(1,3,1)
plt.title('standardscaler')
sns.boxplot(x=d['math score_ss'],data=d)

plt.subplot(1,3,2)
plt.title('before')
sns.boxplot(x=d['math score'],data=d)

plt.subplot(1,3,3)
plt.title('minmaxscaler')
sns.boxplot(x=d['math score_mm'],data=d)


plt.show()
```

```python
from sklearn.preprocessing import FunctionTransformer
import numpy as np
ft=FunctionTransformer(func=np.log1p)
ft.fit(d[['math score']])
d['math score_ft']=ft.transform(d[['math score']])
```

```python
plt.figure(figsize=[20,7])
plt.subplot(1,2,1)
plt.title('functiontransformatiin')
sns.boxplot(x=d['math score_ft'],data=d)

plt.subplot(1,2,2)
plt.title('before')
sns.boxplot(x=d['math score'],data=d)
```
