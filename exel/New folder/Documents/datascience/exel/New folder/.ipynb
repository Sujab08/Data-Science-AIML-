```python
import pandas as pd
```

```python
df=pd.read_csv(r"C:\Users\kumar\Downloads\archive\restaurant_sales_data.csv")
```

```python
df["Order Total"].fillna(df["Price"]*df["Quantity"],inplace=True)
```

```python
df["Price"].fillna(df["Order Total"]/df["Quantity"],inplace=True)
```

```python
df["Quantity"].fillna(df["Order Total"]/df["Price"],inplace=True)
```

```python
df["Order Total"].fillna(df["Price"]*df["Quantity"],inplace=True)
```

```python
df["Price"].fillna(df["Order Total"]/df["Quantity"],inplace=True)
```

```python
df["Quantity"].fillna(df["Order Total"]/df["Price"],inplace=True)
```

```python
df["Item"].fillna("T",inplace=True)
```

```python
for i in range(len(df)):
    if df.loc[i,"Item"]=="T":
        k=df.loc[i,"Price"]
        if k==3.0:
            df.loc[i,"Item"]="Orange Juice"
        elif k==4.0:
            df.loc[i,"Item"]="Mashed Potatoes"
        elif k==2.5:
            df.loc[i,"Item"]="Iced Tea"
        elif k==12.0:
            df.loc[i,"Item"]="Pasta Alfredo"
        elif k==18.0:
            df.loc[i,"Item"]="Salmon"
        elif k==6.0:
            df.loc[i,"Item"]="Chocolate Cake"
        elif k==14.0:
            df.loc[i,"Item"]="Vagetable Platter"
        elif k==5.0:
            df.loc[i,"Item"]="Sweet Potato Fries"
        elif k==7.0:
            df.loc[i,"Item"]="Cheesecake"
        elif k==8.0:
            df.loc[i,"Item"]="Chicken Melt"
        elif k==1.0:
            df.loc[i,"Item"]="Water"
        elif k==10.0:
            df.loc[i,"Item"]="Nachos Grande	"
        elif k==20.0:
            df.loc[i,"Item"]="Steak"
        elif k==15.0:
            df.loc[i,"Item"]="Grilled Chicken"
```

```python
df["Payment Method"].fillna(df["Payment Method"].mode(),inplace=True)
```

```python
df.dropna(inplace=True)
```
