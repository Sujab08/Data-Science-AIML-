# K_mean algorithm

```python
x=df[["Temperature","Humidity","Wind Speed","Precipitation (%)","Atmospheric Pressure","UV Index","Visibility (km)"]]
```

```python
list1=[]
```

```python
from sklearn.cluster import KMeans
```

```python
for i in range(1,21):
    km=KMeans(n_clusters=i,init='random')
    km.fit(x)
    list1.append(km.inertia_)
```

```python
list1
```

```python
list2=[]
```

```python
for i in range(1,21):
    km1=KMeans(n_clusters=i,init='k-means++')
    km1.fit(x)
    list2.append(km1.inertia_)
```

```python
list2
```

```python
plt.figure(figsize=(15,7))
plt.subplot(1,2,1)
plt.plot([i for i in range(1,21)],list2,marker="^")
plt.xlabel("number of cluster")
plt.title("k_mean")
plt.xticks([i for i in range(1,21)])
plt.grid(axis="x")
plt.subplot(1,2,2)
plt.plot([i for i in range(1,21)],list1,marker="^")
plt.xlabel("number of cluster")
plt.title("random")
plt.xticks([i for i in range(1,21)])
plt.grid(axis="x")
plt.show()
```

```python
kmm=KMeans(n_clusters=4)
```
