```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv(r"C:\Users\kumar\Downloads\Placements_Dataset.csv (1).zip")
```

```python
df.head()
```

```python
dff=df.copy()
```

```python
dff.head()
```

```python
dff.groupby('Branch of Engineering')['Placement Package'].max()
```

```python
dff['Knows ML']=dff['Knows ML'].fillna(dff['Knows ML'].mode()[0])
dff['Knows Python']=dff['Knows Python'].fillna(dff['Knows Python'].mode()[0])
dff['Knows JavaScript']=dff['Knows JavaScript'].fillna(dff['Knows JavaScript'].mode()[0])
dff['Knows HTML']=dff['Knows HTML'].fillna(dff['Knows HTML'].mode()[0])
dff['Knows CSS']=dff['Knows CSS'].fillna(dff['Knows CSS'].mode()[0])
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
dff['Knows MLL']=le.fit_transform(dff['Knows ML'])
```

```python
dff['Knows DSAA']=le.fit_transform(dff['Knows DSA'])
dff['Knows Pythonn']=le.fit_transform(dff['Knows Python'])
dff['Knows JavaScriptt']=le.fit_transform(dff['Knows JavaScript'])
dff['Knows HTMLL']=le.fit_transform(dff['Knows HTML'])
dff['Knows CSSS']=le.fit_transform(dff['Knows CSS'])
dff['Knows Crickett']=le.fit_transform(dff['Knows Cricket'])
dff['Knows Dancee']=le.fit_transform(dff['Knows Dance'])
dff['Fest']=le.fit_transform(dff['Participated in College Fest'])
dff['Club']=le.fit_transform(dff['Was in Coding Club'])
```

```python
dff.describe()
```

```python
dff.head()
```

# scaling

```python
from sklearn.preprocessing import StandardScaler
ss=StandardScaler()
ss.fit(dff[['No. of DSA questions']])
```

```python
dff['No. of DSA questions_ss']=pd.DataFrame(ss.transform(dff[['No. of DSA questions']]),columns=['X'])
```

```python
dff['CGPA_ss']=pd.DataFrame(ss.fit_transform(dff[['CGPA']]),columns=['Y'])
```

```python
dff.drop(columns="CGPA_ss",inplace=True)
```

```python
from sklearn.preprocessing import MinMaxScaler
ms=MinMaxScaler()
dff['CGPA_ss']=pd.DataFrame(ms.fit_transform(dff[['CGPA']]),columns=['X'])
```

```python
dff.head()
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
lr=LinearRegression()
X=dff[['No. of DSA questions_ss','CGPA','Knows MLL','Knows DSAA','Knows Pythonn','Knows JavaScriptt','Knows HTMLL','Knows CSSS','No. of backlogs']]
y=dff['Placement Package']
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=42)
lr.fit(X_train,y_train)

```

```python
from sklearn.metrics import mean_squared_error,mean_absolute_error,r2_score
result=lr.predict(X_test)
print("mse:",mean_squared_error(y_test,result))
print("mae:",mean_absolute_error(y_test,result))
print("rmse:",np.sqrt(mean_squared_error(y_test,result)))Location
print("R2score:",r2_score(y_test,result))
```
