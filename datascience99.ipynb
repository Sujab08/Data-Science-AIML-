```python
y=torch.tensor([2,4,6,8])
x=torch.tensor([1,2,3,4])
m=torch.tensor(0.0)
c=torch.tensor(0.0)
```

```python
y_pre=m*x+c
```

```python
loss=((y_pre-y)**2).mean()
```

```python
print(y_pre)
print(y)
print(loss)
```
