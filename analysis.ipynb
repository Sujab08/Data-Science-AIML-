```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

```python
df=pd.read_csv(r"C:\Users\kumar\Downloads\Master_Global_Budgets_Historical.csv")
df
```

```python
helth=df.groupby('Country')['Health_Percentage'].mean().sort_values(ascending=False)
```

```python
plt.figure(figsize=(20,10))
plt.plot(helth.index,helth.values,color='red')
plt.grid()
plt.xticks(rotation=45)
```
