```python
for i in range(0,df.shape[0]):
    list2=[]
    for colum in colums:
        if(type(df[colum][i])==str):
            list2.append(df[colum][i])
    list1.append(list2)
```

```python
market=[]
for i in list1:
    for j in i:
        market.append(j)
```

```python
#market
```

```python
import collections
```

```python
p=collections.Counter(market)
```

```python
se={"product":p.keys(),"values":p.values()}
d=pd.DataFrame(se)
```

```python
d.sort_values(by="values",ascending=False)
```
