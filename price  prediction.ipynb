```python
from sklearn.model_selection import train_test_split
```

```python
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
```

```python
x=df[["Mileage1","Year1","Condition1","Model1","Transmission1","Fuel Type1","Brand1","Engine Size"]]
y=df["Price"]
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,random_state=42,test_size=0.2)
```

```python
from sklearn.linear_model import LinearRegression
lr=LinearRegression()
```

```python
lr.fit(x_train,y_train)
```

```python
lr.score(x_train,y_train),lr.score(x_test,y_test)
```

```python
p=mean_squared_error(y_test,lr.predict(x_test))
```

```python
mean_absolute_error(y_test,lr.predict(x_test)),r2_score(y_test,lr.predict(x_test))
```

```python
np.sqrt(p),p
```

```python
from sklearn.tree import DecisionTreeRegressor
dc=DecisionTreeRegressor(max_depth=90,splitter='best')
```

```python
dc.fit(x_train,y_train)
```

```python
dc.score(x_test,y_test)*100,dc.score(x_train,y_train)
```

```python
np.sqrt(mean_squared_error(y_test,dc.predict(x_test)))
```
