```python
w=torch.tensor(0.0,requires_grad=True)
x=torch.tensor([1.,2.,3.,4.])
y_true=torch.tensor([2.,4.,6.,8.])
lr=0.1
epochs=10
for epoch in range(epochs):
    y_pre=w*x
    loss=((y_pre-y_true)**2).mean()
    loss.backward()
    with torch.no_grad():
        w-=lr*w.grad
    print(f"epoch{epoch+1}:w{w.item():.4f},loss={loss.item():.4f}")
```
